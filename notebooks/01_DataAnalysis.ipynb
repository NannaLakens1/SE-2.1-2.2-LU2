```python
# Cell 1: Import libraries
import pandas as pd
import numpy as np
```

```python
# Cell 2: Create dataset of 10 students with 8 features (some NaN values)
data = {
    "Student": ["Alice", "Bob", "Charlie", "David", "Eva", "Frank", "Grace", "Hannah", "Ian", "Jack"],
    "Math": [90, 85, np.nan, 78, 92, 88, 76, np.nan, 81, 89],
    "Physics": [80, np.nan, 75, 70, 95, 88, 79, 85, 82, np.nan],
    "Chemistry": [np.nan, 78, 82, 75, 88, 91, 80, 77, 85, 90],
    "Biology": [70, 68, 75, np.nan, 80, 72, 78, 85, np.nan, 88],
    "English": [85, 80, 88, 90, 82, np.nan, 75, 80, 79, 87],
    "History": [78, 85, 80, 88, 90, 86, np.nan, 82, 79, 84],
    "Geography": [82, 88, 79, 85, np.nan, 80, 76, 78, 81, 83],
    "Computer": [np.nan, 95, 88, 92, 90, 85, 87, 89, 91, 94]
}

df = pd.DataFrame(data)
df
```

```python
# Cell 3: Detect missing values
df.isnull().sum()
```

```python
# Cell 4: Fill missing values with 0
df_filled = df.fillna(0)
df_filled
```

```python
# Cell 5: Replace NaN with custom value (-99)
df_replaced = df.replace(np.nan, -99)
df_replaced
```

```python
# Cell 6: Drop rows with any missing value
df_drop_any = df.dropna(axis=0, how='any') 
df_drop_any
```

```python
# Cell 7: Drop duplicate rows
df_duplicates = df.copy()
df_duplicates.loc[10] = df_duplicates.loc[0]  # add a duplicate row
df_duplicates.drop_duplicates(inplace=True)
df_duplicates
```

```python
# Cell 8: Change datatype
df["Math"] = df["Math"].astype(float)
df["Student"] = df["Student"].astype(str)
df.dtypes
```

```python
columns = ["Math", "Physics", "Chemistry", "Biology",
           "English", "History", "Geography", "Computer"]

stats = df[columns].agg(["mean", "median", "std"]).T

stats
```

```python
# Cell 9: Add a new column "Total" as sum of all subjects
df["Total"] = df[["Math","Physics","Chemistry","Biology","English","History","Geography","Computer"]].sum(axis=1, skipna=True)
df
```

```python
# Cell 10: Max Absolute Scaling (normalize data between [-1,1])
max_scaled = df.copy()
for col in ["Math","Physics","Chemistry","Biology","English","History","Geography","Computer"]:
    max_scaled[col] = max_scaled[col] / max_scaled[col].abs().max()
max_scaled
```

```python
# Cell 11: Z-Score Normalization (mean=0, std=1)
z_scaled = df.copy()
for col in ["Math","Physics","Chemistry","Biology","English","History","Geography","Computer"]:
    z_scaled[col] = (z_scaled[col] - z_scaled[col].mean()) / z_scaled[col].std()
z_scaled
```

The Tips dataset contains information about restaurant bills and tips. Each row represents one dining party and includes the total bill, tip amount, day, time (lunch or dinner), party size, sex, and whether the customer was a smoker.

It is commonly used to explore relationships between variables, for example: Does a higher total bill result in a higher tip?

```python
import seaborn as sns
import matplotlib.pyplot as plt

tips = sns.load_dataset('tips')

# Select only the numerical columns
numeric_cols = tips.select_dtypes(include='number')

# Calculate the correlation matrix
corr = numeric_cols.corr()

# Show the heatmap
sns.heatmap(corr, annot=True, cmap='coolwarm')
plt.title("Heatmap")
plt.show()
```

```python
import pandas as pd
import numpy as np

# Create a sample dataset with various data types and some missing values
data = {
    "kolomA": [10, 20, np.nan, 40, 20, 20],                
    "kolomB": ["ok", "n/a", "-", "goed", "ok", "n/a"],      
    "datum": ["2024-01-10", "2024/02/05", "invalid", None, "2024-05-01", "2024-05-01"],  
    "categorie": ["A", "B", "A", "B", "C", "A"],            
    "score": [85, 60, 92, 75, 88, 85],                     
    "waarde": [50, 40, 30, 60, 80, 50],
    "totaal": [100, 80, 60, 120, 160, 100],
    "old_name": ["x", "y", "z", "x", "y", "z"]              
}

df = pd.DataFrame(data)

# Double row intentionally
df = pd.concat([df, df.iloc[[1]]], ignore_index=True)

df
```

```python
# Count the number of missing values in each column
df.isnull().sum()
```

```python
# Simply fill missing values in "kolomA" with the median of that column
df["kolomA"] = df["kolomA"].fillna(df["kolomA"].median())
df
```

```python
# Remove duplicates
df = df.drop_duplicates()
df
```

```python
# typeconversion
df["datum"] = pd.to_datetime(df["datum"], errors="coerce")
df["categorie"] = df["categorie"].astype("category")
df
```

```python
# Delete empty columns (if any)
df = df.dropna(axis=1, how="all")
df
```

```python
# Rename a column, create a new column based on a condition, and create a new column as a ratio of two existing columns
df = df.rename(columns={"old_name":"new_name"})
df["label"] = "laag"
df.loc[df["score"] > 80, "label"] = "hoog"
df["ratio"] = df["waarde"]/df["totaal"]
df
```
