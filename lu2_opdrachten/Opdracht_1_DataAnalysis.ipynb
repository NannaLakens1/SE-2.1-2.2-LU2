```python
# 1. Data verkennen en opschonen
# - Importeer de dataset
# - Controleer op missing values, NaN’s, duplicaten en onmogelijke waarden.
# - Maak de data schoon en documenteer jullie keuzes.

import pandas as pd
import numpy as np
import seaborn as sns
import matplotlib.pyplot as plt
lm = pd.read_csv('../lu2_datasets/Leescatalogus_Leesmateriaal.csv', sep=';')
#print(lm.shape)
#lm.head()
```

```python
# aantal unieke en missende waarden en datatypes
print("Aantal rijen en kolommen:", lm.shape)
overzicht = pd.DataFrame({
    "Aantal unieke waarden": lm.nunique(),
    "Aantal missende waarden": lm.isna().sum(),
    "Datatype": lm.dtypes,
    "Aantal dubbele waarden": lm.apply(
        lambda kolom: kolom.duplicated().sum()
    )
})

display(overzicht)
```

```python
# Voor bepaalde kolommen alle unieke waarden bekijken om zo inconsistenties te vinden.
kolommen = ["Taal", "Type_materiaal", "Doelgroep", "Interessecluster", "Beschikbaar_formaat", "Administratieve_prioriteit", "Gegevensstatus"]

waarden = pd.DataFrame({
    kolom: pd.Series(lm[kolom].dropna().unique())
    for kolom in kolommen
})

display(waarden)
```

**Acties**

```python
# de unnamed kolom verwijderen, want hij heeft nergens waarden
lm = lm.drop(columns=['Unnamed: 3'])

# Data Consistent maken:

lm["Type_materiaal"] = (
    lm["Type_materiaal"]
    .str.strip()
    .str.lower()
)

lm["Administratieve_prioriteit"] = (
    lm["Administratieve_prioriteit"]
    .str.strip()
    .str.lower()
)

lm["Beschikbaar_formaat"] = (
    lm["Beschikbaar_formaat"]
    .str.strip()
    .str.lower()
)

lm["Interessecluster"] = (
    lm["Interessecluster"]
    .str.strip()
    .str.lower()
)

lm["Doelgroep"] = (
    lm["Doelgroep"]
    .str.replace(" ","", regex=False)
    .str.lower()
)

lm["Taal"] = lm["Taal"].replace({
    "NED": "Nederlands",
    "NL": "Nederlands",
    "nederlands": "Nederlands"
})

lm["Interne_batchcode"] = lm["Interne_batchcode"].replace({
    "batch-2": "B2",
})

lm["Doelgroep_extra"] = (
    lm["Doelgroep_extra"]
    .str.replace(" ","", regex=False)
    .str.lower()
)

lm["Aandeel_lange_woorden_pct"] = (
    lm["Aandeel_lange_woorden_pct"].str.replace("%", "", regex=False)
)
lm["Aandeel_lange_woorden_pct"] = lm.Aandeel_lange_woorden_pct.astype(float)
```

```python
# Leesmateriaal waar een url bij Themas staat
lm.loc[
    lm["Themas_boek"].str.contains(r"https?://|www\.", na=False, case=False),
    ["Titel", "Themas_boek", "Bron_url"]
]
```
