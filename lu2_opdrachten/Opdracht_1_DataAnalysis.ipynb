**Acties**

```python
# de unnamed kolom verwijderen, want hij heeft nergens waarden
lm = lm.drop(columns=['Unnamed: 3'])

# Data Consistent maken:

lm["Type_materiaal"] = (
    lm["Type_materiaal"]
    .str.strip()
    .str.lower()
)

lm["Administratieve_prioriteit"] = (
    lm["Administratieve_prioriteit"]
    .str.strip()
    .str.lower()
)

lm["Beschikbaar_formaat"] = (
    lm["Beschikbaar_formaat"]
    .str.strip()
    .str.lower()
)

lm["Interessecluster"] = (
    lm["Interessecluster"]
    .str.strip()
    .str.lower()
)

lm["Doelgroep"] = (
    lm["Doelgroep"]
    .str.replace(" ","", regex=False)
    .str.lower()
)

lm["Taal"] = lm["Taal"].replace({
    "NED": "Nederlands",
    "NL": "Nederlands",
    "nederlands": "Nederlands"
})

lm["Interne_batchcode"] = lm["Interne_batchcode"].replace({
    "batch-2": "B2",
})

lm["Doelgroep_extra"] = (
    lm["Doelgroep_extra"]
    .str.replace(" ","", regex=False)
    .str.lower()
)

lm["Auteur"] = lm["Auteur"].replace({
    "Noami Alderman": "Naomi Alderman",
    "H. Mulisch": "Harry Mulisch",
    "Annejet vd Zijl": "Annejet van der Zijl"
})

# ook bronnen weggehaald
lm["Trefwoorden"] = (
    lm["Trefwoorden"]
    .str.replace(r"\s*[,/|;]\s*", ";", regex=True)
    .str.strip()
    .str.replace(r"\S*\.(?:nl|com)\S*", "", regex=True)
    .replace(["", "nan"], pd.NA)
)

# Bronnen weggehaald bij Onderwerp
lm["Onderwerp"] = lm["Onderwerp"].str.replace(r"\S*\.(?:nl|com)\S*", "", regex=True).replace(["", "nan"], pd.NA)

# Bronnen weggehaald, constistente naamgeving (er was ; , / |), en consistente null
lm["Themas_boek"] = (
    lm["Themas_boek"]
    .str.replace(r"\S*\.(?:nl|com)\S*", "", regex=True, case=False)
    .str.replace(r"\s*[,/|;]\s*", ";", regex=True)
    .str.strip(" ;")
    .replace(["", "nan"], pd.NA)
)

# Aantal pagina's opschonen: - en 0 waarden weg, consistente NaN en int in plaats van float. 
lm["Aantal_paginas"] = lm["Aantal_paginas"].mask(
    lm["Aantal_paginas"] <= 0
).astype("Int64")

# bij Aandeel_lange_woorden_pct de '%' verwijderd en een float gemaakt
lm["Aandeel_lange_woorden_pct"] = (
    lm["Aandeel_lange_woorden_pct"].str.replace("%", "", regex=False)
)
lm["Aandeel_lange_woorden_pct"] = lm.Aandeel_lange_woorden_pct.astype(float)

# dubbele Materiaal_id's automatisch corrigeren
duplicaten = lm["Materiaal_id"].duplicated(keep="first")
for index in lm.index[duplicaten]:
    huidig_id = lm.loc[index, "Materiaal_id"]
    nummer = int(huidig_id.replace("LM-", ""))

    while f"LM-{nummer + 1:04d}" in lm["Materiaal_id"].values:
        nummer += 1

    lm.loc[index, "Materiaal_id"] = f"LM-{nummer + 1:04d}"

# doelgroep en doelgroep_extra combineren en doelgroep_extra verwijderen
def combineer_doelgroep(rij):
    waarden = set()

    for kolom in ["Doelgroep", "Doelgroep_extra"]:
        waarde = rij[kolom]

        if pd.notna(waarde):
            waarden.update(waarde.lower().split("/"))

    volgorde = ["mbo", "nt2"]
    resultaat = [waarde for waarde in volgorde if waarde in waarden]

    return "/".join(resultaat) if resultaat else pd.NA
lm["Doelgroep"] = lm.apply(combineer_doelgroep, axis=1)
lm = lm.drop(columns=['Doelgroep_extra'])

# Cataloguscode verwijderen, want dit is hetzelfde als Materiaal_id
#lm = lm.drop(columns=['Cataloguscode'])

# Genre_kopie verwijderen, want deze bevat geen aanvullende informatie op Genre
lm = lm.drop(columns=['Genre_kopie'])

# Gegevensstatus aangepast: nu is het de status beschikbaar, niet beschikbaar en uitgeleend
lm["Gegevensstatus"] = lm["Gegevensstatus"].replace({
    "Aangevuld (deels gesimuleerd)": "beschikbaar",
    "gecontroleerd": "niet beschikbaar",
    "In behandeling": "uitgeleend"
}).fillna("niet beschikbaar")

# twee gecontroleerde publicatiejaren corrigeren
lm.loc[lm["Titel"] == "De roadtrip", "Publicatiejaar"] = "2025"
lm.loc[lm["Titel"] == "Mijn naam is Edino", "Publicatiejaar"] = "2025"
# Overige ongeldige jaren als NaN
lm["Publicatiejaar"] = lm["Publicatiejaar"].replace({
    "2099": pd.NA,
    "onbekend": pd.NA,
    "nan": pd.NA
})

 
```
