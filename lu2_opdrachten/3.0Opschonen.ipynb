```python
import re
def vind_themas(omschrijving):
    if pd.isna(omschrijving):
        return []

    gevonden = []

    for thema in themas:
        patroon = r"\b" + re.escape(thema) + r"\b"

        if re.search(patroon, str(omschrijving), re.IGNORECASE):
            gevonden.append(thema)

    return ";".join(gevonden)


df["Gevonden_themas"] = df["Korte_omschrijving"].apply(vind_themas)
df["Gevonden_themas"] = df["Gevonden_themas"].replace(
    r"^\s*$", pd.NA, regex=True
)
```

```python
import re
import pandas as pd

def vind_genres(omschrijving, bestaand_genre):
    gevonden = []

    # Bestaande genres behouden
    if pd.notna(bestaand_genre):
        gevonden.extend(str(bestaand_genre).split(";"))

    # Genres die we zoeken
    genre_woorden = {
        "thriller": "thriller",
        "roman": "roman",
        "jeugdroman": "roman",
        "novel": "roman",
        "novelle": "roman",
        "humor": "comedy"
    }

    if pd.notna(omschrijving):
        for woord, genre in genre_woorden.items():
            patroon = r"\b" + re.escape(woord) + r"\b"

            if re.search(patroon, str(omschrijving), re.IGNORECASE):
                gevonden.append(genre)

    # Opschonen en dubbele genres verwijderen
    resultaat = []
    for genre in gevonden:
        genre = genre.strip().lower()

        if genre and genre not in resultaat:
            resultaat.append(genre)

    return ";".join(resultaat)


df["Genre"] = df.apply(
    lambda row: vind_genres(
        row["Korte_omschrijving"],
        row["Genre"]
    ),
    axis=1
)
df["Genre"] = df["Genre"].replace(
    r"^\s*$", pd.NA, regex=True
)
```

```python
def verwijder_dubbele_themas(rij):
    # Haal de bestaande thema's op
    themas_boek = set(
        woord.strip().lower()
        for woord in str(rij["Themas_boek"]).split(";")
        if woord.strip() and woord.strip().lower() != "nan"
    )

    # Haal de gevonden thema's op
    gevonden_themas = str(rij["Gevonden_themas"]).split(";")

    # Bewaar alleen thema's die nog niet in Themas_boek staan
    unieke_themas = []

    for thema in gevonden_themas:
        thema = thema.strip()

        if thema and thema.lower() not in themas_boek and thema.lower() != "nan":
            if thema.lower() not in [t.lower() for t in unieke_themas]:
                unieke_themas.append(thema)

    return ";".join(unieke_themas)


df["Gevonden_themas"] = df.apply(verwijder_dubbele_themas, axis=1)
df["Gevonden_themas"] = df["Gevonden_themas"].replace(
    r"^\s*$", pd.NA, regex=True
)
```

```python
df["Themas_boek"] = (
    df["Themas_boek"]
    .str.lower()
)
df["Gevonden_themas"] = (
    df["Gevonden_themas"]
    .str.lower()
)
df["Trefwoorden"] = (
    df["Trefwoorden"]
    .str.lower()
)
```
