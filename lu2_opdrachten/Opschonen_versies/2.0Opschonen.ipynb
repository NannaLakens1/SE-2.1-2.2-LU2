```python
# de Materiaal_id kolom verwijderen, want dit is hetzelfde als Catalogus_code, beide unieke id, en dezelfde nummerieke getallen
df = df.drop(columns=['Materiaal_id'])

df.loc[df["Gem_woordlengte"] > 15, "Gem_woordlengte"] = float("nan")
```

```python
df.to_csv("Leescatalogus_Opgeschoond.csv", index=False)
```

```python
import pandas as pd

def vergelijk_themas(rij):
    # Zet beide kolommen om naar sets met losse woorden
    themas = set(
        woord.strip().lower()
        for woord in str(rij["Themas_boek"]).split(";")
        if woord.strip() and woord.strip().lower() != "nan"
    )

    trefwoorden = set(
        woord.strip().lower()
        for woord in str(rij["Trefwoorden"]).split(";")
        if woord.strip() and woord.strip().lower() != "nan"
    )

    return themas == trefwoorden


df["Themas_gelijk"] = df.apply(vergelijk_themas, axis=1)

# Toon de rijen waarbij de waarden niet overeenkomen
verschillen = df[df["Themas_gelijk"] == False]

print(verschillen[["Themas_boek", "Trefwoorden"]].to_string())
```

```python
import pandas as pd

def onderwerpen_in_themas(rij):
    themas = set(
        woord.strip().lower()
        for woord in str(rij["Themas_boek"]).split(";")
        if woord.strip() and woord.strip().lower() != "nan"
    )

    onderwerpen = set(
        woord.strip().lower()
        for woord in str(rij["Onderwerp"]).split(";")
        if woord.strip() and woord.strip().lower() != "nan"
    )

    return onderwerpen.issubset(themas)


df["Onderwerpen_in_themas"] = df.apply(onderwerpen_in_themas, axis=1)

# Toon alleen de rijen waarbij onderwerpen niet volledig in Themas_boek staan
verschillen = df[df["Onderwerpen_in_themas"] == False]

print(verschillen[["Themas_boek", "Onderwerp"]].to_string())
```

```python
print("Aantal rijen:", len(df))

# Themas_boek vergelijken met Trefwoorden
print("\n--- Themas_boek vs. Trefwoorden ---")
print("Aantal overeenkomsten:", (df["Themas_gelijk"] == True).sum())
print("Aantal verschillen:", (df["Themas_gelijk"] == False).sum())

# Onderwerpen controleren tegen Themas_boek
print("\n--- Onderwerpen vs. Themas_boek ---")
print("Aantal overeenkomsten:", (df["Onderwerpen_in_themas"] == True).sum())
print("Aantal verschillen:", (df["Onderwerpen_in_themas"] == False).sum())
```
