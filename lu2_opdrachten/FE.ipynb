```python
import pandas as pd
df = pd.read_csv("../lu2_datasets/Leescatalogus_Opgeschoond.csv")
```

```python
kolommen = [
    "korte_omschrijving",
    "themas_boek",
    "trefwoorden",
    "onderwerp",
    "genre",
    "doelgroep",
    "niveau",
    "type_materiaal"
]

df["gecombineerde_tekst"] = (
    df[kolommen]
    .fillna("")
    .astype(str)
    .agg(" ".join, axis=1)
    .str.replace(r"\s+", " ", regex=True)
    .str.strip()
)
```

```python
pd.set_option("display.max_colwidth", None)
display(df[["titel", "gecombineerde_tekst", "cataloguscode"]])
```

```python
df.to_csv("Leescatalogus_gecombineerde_tekst.csv", index=False)
```
